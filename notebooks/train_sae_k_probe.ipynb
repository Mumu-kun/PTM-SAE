# PTM-SAE TopK k probe: explained variance versus sparsity (stage 0.5)

The same notebook as `train_sae.ipynb`, run by default on `sweeps/topk_k_probe.yaml`: five short TopK runs (width 4,096, k = 64 to 1,024, 3,000 steps) that show where explained variance reaches the 85% gate. Only the two parameter defaults below differ (`PTM_SAE_SWEEP_SPEC`, `PTM_SAE_BENCHMARK`); `tests/test_notebooks.py` fails if the cells drift apart.

# (training notebook description follows)
### Salvage-proof: guarded cells, training as a logged subprocess, checkpoints kept on any failure

**Modes** (`PTM_SAE_MODE` env var): `smoke` is a 200-step run on the small 8M pilot activations
(about a minute; proves the notebook, not a result); `real` trains `TRAIN_PROFILE`'s full config.

1. **Environment, repo, install, preflight**: GPU, disk, internet, secrets (names only) checked up front; a missing W&B key just turns W&B off.
2. **GPU speed probe** (real mode): tokens/sec on this machine's GPU before committing `total_steps`.
3. **Train**: streams `discovery_train`, evaluates on `discovery_val`, writes `latest`/`best` checkpoints. Runs as a subprocess with a log; a session-limit stop keeps `latest` for `--resume-from`.
4. **Checkpoint readback** and **post-training diagnostics** (W&B charts and verdict), each independent.
5. **Run summary**: always printed, even after failures.

Runs identically on a Kaggle kernel and a Linux GPU box (the box is preferred; use Kaggle only when
it is unavailable; a run never uses both).

```python
# 2. Repository Setup (same on a local PC, a persistent remote box, Kaggle and Colab)
# Finds the checkout by walking up from the kernel's working directory, so it also works when this
# notebook is opened from notebooks/. Only a runtime with no checkout (fresh Kaggle/Colab) clones;
# an existing checkout is fast-forwarded (local edits are auto-stashed, never discarded) on its
# current branch -- switch branches yourself with `git checkout` in a terminal.
import os
import subprocess
import sys
from pathlib import Path

BRANCH = os.environ.get("PTM_SAE_BRANCH", "main")  # used only when cloning a fresh runtime
PULL = os.environ.get("PTM_SAE_PULL", "1") == "1"  # fast-forward an existing checkout (PTM_SAE_PULL=0 to skip)
REPO = "github.com/Mumu-kun/PTM-SAE.git"

here = Path.cwd().resolve()
root = next(
    (p for p in (here, *here.parents) if (p / "pyproject.toml").is_file() and (p / "src/ptm_sae").is_dir()),
    None,
)

if root is None:
    token = os.environ.get("GH_TOKEN")  # private repo: also read Kaggle/Colab secrets
    try:
        from kaggle_secrets import UserSecretsClient

        token = token or UserSecretsClient().get_secret("GH_TOKEN")
    except Exception:
        pass
    try:  # an API-pushed kernel has no attached Secrets: read the private secrets dataset instead
        import json

        found = [p for pattern in ("*/ptm_sae_secrets.json", "*/*/ptm_sae_secrets.json") for p in Path("/kaggle/input").glob(pattern)]
        token = token or json.loads(found[0].read_text(encoding="utf-8")).get("GH_TOKEN")
    except Exception:
        pass
    try:
        from google.colab import userdata

        token = token or userdata.get("GH_TOKEN")
    except Exception:
        pass

    auth = f"x-access-token:{token}@" if token else ""
    root = (Path("/kaggle/working") if Path("/kaggle").exists() else here) / "PTM-SAE"
    print(f"Cloning {BRANCH} into {root}...")
    try:
        subprocess.run(["git", "clone", "--branch", BRANCH, f"https://{auth}{REPO}", str(root)], check=True, capture_output=True)
    except subprocess.CalledProcessError:
        raise RuntimeError(f"git clone of branch {BRANCH!r} failed: check the branch name and that the GH_TOKEN secret is attached to this notebook") from None
elif PULL:
    print(f"Fast-forwarding existing checkout at {root}...")
    subprocess.run(["git", "pull", "--ff-only", "--autostash"], cwd=root, check=False)

os.chdir(root)
if str(root / "src") not in sys.path:
    sys.path.insert(0, str(root / "src"))

from ptm_sae import runtime

print(f"Platform: {runtime.detect_platform()} | repo: {root} | data root: {runtime.resolve_data_root()}")
```

```python
# 3. Install Dependencies (single-sourced from pyproject.toml)
# Local/remote boxes manage their environment with `uv sync --extra notebook` in a terminal (locked
# versions, CUDA torch build kept intact), so there this cell only checks it. Kaggle/Colab ship a
# CUDA-matched torch/transformers, so there everything else is pip-installed (retried: a flaky mirror
# must not end a long run) and those two are left alone. A failure is recorded below, not raised.
import importlib.metadata
import subprocess
import sys
import time

EXTRAS = ()
INSTALL_OK = True
# Packages this kernel already imported must not be changed by pip underneath it (the running process
# would be half-old, half-new and later imports fail): remember their versions to check afterwards.
IMPORTED_BEFORE = {n: importlib.metadata.version(n) for n in ("numpy", "pandas", "pyarrow", "scipy", "torch") if n in sys.modules}

if runtime.detect_platform() == "local":
    missing = runtime.missing_requirements(extras=EXTRAS)
    INSTALL_OK = not missing
    if missing:
        print(f"Missing {missing}: run `uv sync --extra notebook` in a terminal, then restart the kernel.")
else:
    requirements = runtime.cloud_requirements(extras=EXTRAS)
    for attempt in range(1, 4):
        deps = subprocess.run([sys.executable, "-m", "pip", "install", "-q", *requirements], capture_output=True, text=True)
        local = subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", ".", "--no-deps", "--ignore-requires-python"], capture_output=True, text=True)
        if deps.returncode == 0 and local.returncode == 0:
            break
        print(f"pip attempt {attempt} failed: {(deps.stderr + local.stderr)[-400:]}")
        time.sleep(10 * attempt)
    else:
        INSTALL_OK = False
changed = [n for n, version in IMPORTED_BEFORE.items() if importlib.metadata.version(n) != version]
if changed:
    INSTALL_OK = False
    print(f"RESTART THE KERNEL: {changed} were imported before pip changed them; later imports would fail.")
print("dependencies ready:", INSTALL_OK)
```

```python
# 4. Run mode, parameters, run state and preflight
import os
import sys
from pathlib import Path

MODE = os.environ.get("PTM_SAE_MODE", "real")  # "smoke": tiny run proving the notebook end to end; "real": the full run
if MODE not in ("smoke", "real"):
    print(f"Unknown PTM_SAE_MODE={MODE!r}; using 'smoke'")
    MODE = "smoke"

TRAIN_PROFILE = os.environ.get("PTM_SAE_PROFILE", "topk")  # real mode: 'topk' | 'jumprelu' | 'batchtopk' | 'gated' -> configs/train_<profile>_baseline.yaml
QUICK_DIAGNOSTIC_RUN = os.environ.get("PTM_SAE_QUICK", "0") == "1"  # real mode: shrink to a short run first (step counts scaled proportionally)
QUICK_DIAGNOSTIC_STEPS = 2000
SWEEP_SPEC = os.environ.get("PTM_SAE_SWEEP_SPEC", "sweeps/topk_k_probe.yaml") or None  # e.g. "sweeps/topk_sweep.yaml": run that sweep instead of one training
BENCHMARK = os.environ.get("PTM_SAE_BENCHMARK", "0") == "1"  # sweeps: measure how many runs to share a GPU before launching
SWEEP_PARALLEL = None  # set by the benchmark cell (or here): concurrent runs across all GPUs
PARTITION_FOLDERS = None  # None: whatever the profile's config says (the 650M data is per partition, the 8M smoke pilot is mixed); True/False overrides
ACTIVATIONS_DIR = None  # e.g. "/kaggle/input/ptm-sae-activations": an attached Kaggle Dataset, read as is (no Hub download)

DATA_ROOT = runtime.resolve_data_root()
LOG_DIR = DATA_ROOT / "logs"

# Every stage below runs inside `with state.cell(...)`: an exception is recorded and printed, later
# cells that depend on it skip with a reason, and the run carries on, so one silly error can never
# throw away a long run's outputs.
state = runtime.RunState()
with state.cell("install"):
    if not INSTALL_OK:
        raise RuntimeError("dependency install failed (see the install cell output)")

with state.cell("preflight"):
    state.require("install")
    report = runtime.preflight(need_gpu=MODE == "real", need_hub_write=False, min_free_gb=10)
    if not report.ok:
        raise RuntimeError(f"preflight failed: {report.failures()}")
print(f"mode={MODE} profile={TRAIN_PROFILE if MODE == 'real' else 'smoke pilot'} data root={DATA_ROOT}")
```

```python
# 5. GPU speed probe (real mode)
# Synthetic tensors at the real shapes: real ms/step and tokens/sec on this machine's GPU, so
# total_steps in the next cell is not a guess.
with state.cell("speed_probe"):
    state.require("preflight")
    if MODE == "smoke":
        raise runtime.CellSkipped("smoke mode")
    import time

    import torch

    from ptm_sae.models import JumpReLUSAEConfig, JumpReLUSAEModel, TopKSAEConfig, TopKSAEModel

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"device: {device}")

    d_in, batch, n_steps = 1280, 4096, 30

    def bench(model, l0_coef=0.0):
        model = model.to(device)
        opt = torch.optim.Adam(model.parameters(), lr=1e-4)
        x = torch.randn(batch, d_in, device=device)
        for _ in range(3):
            out = model(x)
            loss = out.mse_loss + l0_coef * out.l0
            loss.backward()
            model.remove_decoder_gradient_parallel_component_()
            opt.step()
            opt.zero_grad()
            model.normalize_decoder_()
        if device.type == "cuda":
            torch.cuda.synchronize()
        t0 = time.perf_counter()
        for _ in range(n_steps):
            out = model(x)
            loss = out.mse_loss + l0_coef * out.l0
            loss.backward()
            model.remove_decoder_gradient_parallel_component_()
            opt.step()
            opt.zero_grad()
            model.normalize_decoder_()
        if device.type == "cuda":
            torch.cuda.synchronize()
        elapsed = (time.perf_counter() - t0) / n_steps
        return elapsed, batch / elapsed

    for width in (4096, 10240):
        m = TopKSAEModel(TopKSAEConfig(d_in=d_in, d_hidden=width, k=64))
        per_step, tps = bench(m)
        print(f"TopK      width={width:>6}  {per_step * 1000:7.2f} ms/step   {tps:>10,.0f} tok/s")

        m2 = JumpReLUSAEModel(JumpReLUSAEConfig(d_in=d_in, d_hidden=width))
        per_step2, tps2 = bench(m2, l0_coef=1e-3)
        print(f"JumpReLU  width={width:>6}  {per_step2 * 1000:7.2f} ms/step   {tps2:>10,.0f} tok/s")
```

```python
# 6. Training config: the profile's YAML anchored under the data root, plus soft fallbacks from preflight
with state.cell("config"):
    state.require("preflight")
    import yaml

    from ptm_sae.training.config import SAETrainingConfig

    CONFIG_PATH = "configs/train_smoke_pilot.yaml" if MODE == "smoke" else f"configs/train_{TRAIN_PROFILE}_baseline.yaml"
    config = SAETrainingConfig.from_yaml(CONFIG_PATH).with_data_root()
    if PARTITION_FOLDERS is not None:
        config.partition_folders = PARTITION_FOLDERS
    import torch

    GPU_ARGS = ["--gpus", ",".join(str(i) for i in range(torch.cuda.device_count()))] if torch.cuda.device_count() else []
    if ACTIVATIONS_DIR:
        config.cache_dir, config.remote_repo_id = str(ACTIVATIONS_DIR), None
    if "wandb" in report.degraded:
        config.wandb.enabled = False
        print("W&B off:", report.degraded["wandb"])

    # Short sanity pass: every step/token-derived field scales with total_steps, so a short run is not
    # artificially healthier than a full one (e.g. dead-latent window never reached, warmup never done).
    if QUICK_DIAGNOSTIC_RUN and MODE == "real":
        shrink = QUICK_DIAGNOSTIC_STEPS / config.total_steps
        config.total_steps = QUICK_DIAGNOSTIC_STEPS
        config.warmup_steps = max(1, int(config.warmup_steps * shrink))
        config.eval_interval_steps = max(1, int(config.eval_interval_steps * shrink))
        config.collapse_check_interval_steps = max(1, int(5 * config.eval_interval_steps))
        config.dead_latent_window_tokens = max(1, int(config.dead_latent_window_tokens * shrink))
        config.wandb.run_name = f"{config.wandb.run_name}_quick"
        config.wandb.tags = [*config.wandb.tags, "quick-diagnostic"]
        print(f"[Quick Diagnostic] scaled every step/token field by {shrink:.3f}x")

    RUN_CONFIG = DATA_ROOT / "run_config.yaml"
    RUN_CONFIG.write_text(yaml.safe_dump(config.model_dump(mode="json")))
    print(f"profile: {CONFIG_PATH} | run config: {RUN_CONFIG}")
    print(config.model_dump_json(indent=2))
```

```python
# 6b. Data: bring the corpus tables and both partitions' shards onto this machine in one visible, resumable
# step, so parallel runs never race to download the same shards. An attached Kaggle Dataset needs no download.
with state.cell("data"):
    state.require("config")
    what = "corpus" if ACTIVATIONS_DIR else "all"
    cmd = [sys.executable, "-u", "-m", "ptm_sae.data.sync", "--config", str(RUN_CONFIG), "--what", what]
    code = runtime.run_logged(cmd, LOG_DIR / "data_sync.log", timeout_s=state.time_left())
    if code != 0:
        raise RuntimeError(f"data sync exited with {code}; see {LOG_DIR / 'data_sync.log'} (re-run resumes)")
```

```python
# 7. Train. A subprocess with a log: a crash cannot take the notebook (or the checkpoints) down.
# latest/ and best/ are written as it goes, so even a failed or time-limited run leaves a checkpoint.
with state.cell("train"):
    state.require("data")
    if SWEEP_SPEC:
        raise runtime.CellSkipped(f"sweep mode ({SWEEP_SPEC})")
    cmd = [sys.executable, "-u", "-m", "ptm_sae.training.train", "--config", str(RUN_CONFIG)]
    code = runtime.run_logged(cmd, LOG_DIR / "train.log", timeout_s=state.time_left())
    if code == runtime.TIMEOUT_EXIT_CODE:
        raise RuntimeError("stopped at the session time limit; resume with --resume-from <checkpoint_dir>")
    if code != 0:
        raise RuntimeError(f"training exited with {code}; see {LOG_DIR / 'train.log'}")
```

```python
# 7b. Concurrency benchmark (sweeps): identical short runs at 1, 2, 4 and 6 at once; the smallest level within 5%
# of the best aggregate tokens/sec is used for the sweep, because CPU data loading, not the GPU, is the limit.
with state.cell("benchmark"):
    state.require("data")
    if not SWEEP_SPEC or not BENCHMARK or MODE == "smoke":
        raise runtime.CellSkipped("no sweep, benchmark off, or smoke mode")
    import json

    base_config = yaml.safe_load(Path(SWEEP_SPEC).read_text(encoding="utf-8"))["base_config"]
    bench_out = DATA_ROOT / "benchmark.json"
    cmd = [
        sys.executable, "-u", "-m", "ptm_sae.training.sweep", "--benchmark", "--config", base_config,
        "--levels", "1,2,4,6", "--steps", "300", "--data-root", str(DATA_ROOT),
        "--benchmark-out", str(bench_out), *GPU_ARGS,
    ]  # fmt: skip
    code = runtime.run_logged(cmd, LOG_DIR / "benchmark.log", timeout_s=min(state.time_left(), 3600))
    if code != 0:
        raise RuntimeError(f"benchmark exited with {code}; see {LOG_DIR / 'benchmark.log'} (the sweep uses its default concurrency)")
    SWEEP_PARALLEL = json.loads(bench_out.read_text(encoding="utf-8"))["chosen"]
    print(f"sweep concurrency: {SWEEP_PARALLEL}")
```

```python
# 7c. Sweep: every run of the spec as its own training subprocess, several per GPU. Runs are isolated (one
# failing does not stop the others) and re-running resumes: finished runs are skipped, interrupted ones continue.
with state.cell("sweep"):
    state.require("data")
    if not SWEEP_SPEC:
        raise runtime.CellSkipped("no sweep spec")
    from ptm_sae.training.sweep import SweepState

    spec = yaml.safe_load(Path(SWEEP_SPEC).read_text(encoding="utf-8"))
    cmd = [
        sys.executable, "-u", "-m", "ptm_sae.training.sweep", "--spec", SWEEP_SPEC,
        "--data-root", str(DATA_ROOT), "--timeout-hours", str(state.time_left() / 3600), *GPU_ARGS,
    ]  # fmt: skip
    if SWEEP_PARALLEL:
        cmd += ["--max-parallel", str(SWEEP_PARALLEL)]
    code = runtime.run_logged(cmd, LOG_DIR / "sweep.log", timeout_s=state.time_left() + 600)
    if code != 0:
        raise RuntimeError(f"sweep launcher exited with {code}; see {LOG_DIR / 'sweep.log'}")
    sweep_runs = SweepState(DATA_ROOT / "sweeps" / spec["group"] / "state.json").runs
    for run_name, run_state in sweep_runs.items():
        print(f"{run_name:32s} {run_state.get('status')}")
    unfinished = [name for name, run_state in sweep_runs.items() if run_state.get("status") != "ok"]
    if unfinished:
        raise RuntimeError(f"{len(unfinished)} sweep run(s) did not finish: {unfinished} (re-running resumes them)")
```

```python
# 8. Checkpoint readback: confirms the saved best checkpoint reloads and runs (independent of the train cell's status)
with state.cell("readback"):
    state.require("config")
    import torch

    from ptm_sae.training.train import MODEL_CLASS_BY_TYPE

    best = Path(config.checkpoint_dir) / "best"
    if not best.exists():
        raise runtime.CellSkipped(f"no best checkpoint at {best}")
    reloaded = MODEL_CLASS_BY_TYPE[config.sae_type].from_pretrained(best)
    out = reloaded(torch.randn(4, config.d_in))
    print(f"[Verification] reloaded '{config.sae_type}' from {best}: reconstruction {tuple(out.reconstruction.shape)}, l0 {out.l0.item():.1f}")
```

## Post-training diagnostics
Same charts/verdict/recommendation logic as `notebooks/jumprelu_diagnostic_report.ipynb`
(`ptm_sae.analysis`), applied to the run just trained. They need W&B logging; with W&B off they skip.

```python
# 9. Recover the W&B run id and pull the run's metrics (needs wandb.enabled=true and an API key)
with state.cell("diagnostics_pull"):
    state.require("config")
    import torch

    from ptm_sae.analysis import fetch_run, pull_histogram, pull_scalars

    resume_state = Path(config.checkpoint_dir) / "latest" / "resume_state.pt"
    if not resume_state.exists():
        raise runtime.CellSkipped("no latest checkpoint to read the W&B run id from")
    wandb_run_id = torch.load(resume_state, map_location="cpu", weights_only=False)["wandb_run_id"]
    if wandb_run_id is None:
        raise runtime.CellSkipped("this run had wandb.enabled=false: nothing to pull")

    diag_run = fetch_run(config.wandb.project, wandb_run_id)
    common_keys = [
        "train/decoder_pre_norm_mean", "val/explained_variance", "val/cosine_sim_mean",
        "val/cosine_sim_p10", "val/mse", "val/dead_latent_fraction", "val/alive_latent_jaccard",
    ]
    arch_keys = {
        "jumprelu": ["train/threshold_std", "train/threshold_mean", "train/decoder_norm_mean", "train/l0"],
        "topk": ["train/l0"],
    }.get(config.sae_type, [])
    diag_scalars = pull_scalars(diag_run, common_keys + arch_keys)
    diag_histogram = pull_histogram(diag_run, "val/feature_density_histogram")
    print(f"[Diagnostics] wandb_run_id={wandb_run_id}")
```

```python
# 10. Charts, verdict and recommendations
with state.cell("diagnostics_report"):
    state.require("diagnostics_pull")
    from IPython.display import Image, display

    from ptm_sae.analysis import plot_checks, run_diagnostics

    out_dir = Path(config.checkpoint_dir) / "diagnostic_figures"
    for path in plot_checks(diag_scalars, config.sae_type, out_dir, histogram=diag_histogram):
        display(Image(filename=str(path)))

    result = run_diagnostics(diag_scalars, config.sae_type, diag_run.config)
    display(result.verdict)
    for observed, diagnosis, recommendation in result.recommendations:
        print(f"- {observed}")
        if diagnosis:
            print(f"  diagnosis:   {diagnosis}")
        if recommendation:
            print(f"  recommend:   {recommendation}")
```

```python
# 11. Run summary -- always runs: what completed, what did not, and the files that can be salvaged
try:
    state.print_summary(artifact_dirs=(Path(config.checkpoint_dir) if state.ok("config") else DATA_ROOT / "checkpoints", DATA_ROOT / "sweeps", LOG_DIR))
except Exception as exc:  # noqa: BLE001 -- the summary must never be what fails the run
    print(f"summary failed: {type(exc).__name__}: {exc}")
```
